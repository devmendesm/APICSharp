```python
# 19. Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o angulo entre eles (produto escalar normalizado).
import numpy as np

array1 = np.random.randint(1, 101, 20)
array2 = np.random.randint(1, 101, 20)

# pega o produto escalar com a função np.dot()
produto_escalar = np.dot(array1, array2)

# calcula a norma e o cosseno
norma1 = np.linalg.norm(array1)
norma2 = np.linalg.norm(array2)
cosseno = produto_escalar / (norma1 * norma2)

# a função np.arccos() retorna o arco-cosseno em radianos, ai transforma para graus
angulo = np.arccos(cosseno)
angulo_graus = np.degrees(angulo)

print("Array 1: ", array1)
print("Array 2: ", array2)
print("Produto escalar: ", produto_escalar)
print(f"Ângulo: {angulo_graus} graus" )
```

```python
# 1. Calcule a integral da funcao f(x) = x2 sin(x) no intervalo de 0 a π.
from scipy import integrate
import numpy as np

# utiliza a função np.sin() que representa a função desejada
def f(x):
  return x ** 2 * np.sin(x)

# calcula a integral da função f, começando em 0 e terminando em pi, retornando 2 valores para as variaveis
resultado, erro = integrate.quad(f, 0, np.pi)

print("Integral: ", resultado)
print("Erro estimado: ", erro)
```

```python
# 2. Resolva a equacao diferencial dy dx = x − y, com y(0) = 1, no intervalo de 0 a 10.
from scipy.integrate import solve_ivp
import numpy as np

def equacao(x, y):
  return x - y

# solve_ivp(equação, intervalo, condição_inicial, pontos_para_avaliar)
solucao = solve_ivp(
    equacao,
    [0, 10],
    [1],
    t_eval = np.linspace(0, 10, 100)
)

print("Valores de X: \n", solucao.t)
print("\nValores de Y: \n", solucao.y[0])
```

```python
# 3. Encontre as raızes da funcao f(x) = x 4 − 3x 3 + 2 utilizando SciPy.
from scipy.optimize import root

def f(x):
  return x ** 4 - 3 * x ** 3 + 2

# procura uma raiz partindo de 1
raiz = root(f, 1)

print("Raiz encontrada: ", raiz.x)
```

```python
# 4. Use a funcao de otimizacao de SciPy para maximizar f(x) = −x2 + 4x+ 1.
from scipy.optimize import minimize

def f(x):
  return - x ** 2 + 4 * x + 1

def f_negativa(x):
  return -f(x)

resultado = minimize(f_negativa, 0)
x_max = resultado.x
valor_max = f(x_max)

print("Maximo encontrado em x = ", x_max)
print("Valor maximo = ", valor_max)

```

```python
# 5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).
from scipy.fft import fft
import numpy as np

# cria valores de tempo, criando 500 pontos entre 0 e 1
t = np.linspace(0, 1, 500)

# cria o sinal
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

# calcula a transformada de Fourier
transformada = fft(sinal)

print("Transformada de Fourier: \n", transformada)
```

```python
# 7. Use a funcao de interpolacao cubica de SciPy para interpolar os dados (0, 0),(1, 2),(2, 3),(3, 5),(4, 4).
from scipy.interpolate import interp1d
import numpy as np

# Pontos conhecidos
x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])

# cria a função de interpolação cubica
f = interp1d(x, y, kind = 'cubic')

# novos pontos onde interpolar, para descobrir valores entre os originais
x_novos = np.array([0.5, 1.5, 2.5, 3.5])

# calcula os valores interpolados
y_novos = f(x_novos)

print("Valores interpolados: ", y_novos)
```

```python
# 9. Ajuste uma curva quadratica aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8).
from scipy.optimize import curve_fit
import numpy as np

# Pontos
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

# Função quadrática
def parabola(x, a, b, c):
  return a * x ** 2 + b * x + c

# Ajustar a curva
parametros, _ = curve_fit(parabola, x, y)

print("Parametros [a, b, c]: ", parametros)
```

```python
# 10. Calcule a integral dupla da funcao f(x, y) = xy + x 2 no intervalo [0, 2] para x e [0, 1] para y.
from scipy.integrate import dblquad

# recebe os parametros nessa ordem, pois a dblquad() espera nesse formato
def f(y, x):
  return x * y + x ** 2

resultado, erro = dblquad(
    f,
    0, 2,        # intervalo de x
    lambda x: 0, # significa limite inferior de y = 0
    lambda x: 1  # intervalo de y
)

print("Resultado: ", resultado)
print("Erro estimado: ", erro)
```

```python
# 16. Resolva a equacao diferencial de segunda ordem d2y dx2 +y = 0, com condicoes iniciais y(0) = 1 e y′(0) = 0.
from scipy.integrate import solve_ivp
import numpy as np

def equacao(x, estado):
  y, dy = estado
  return [dy, -y]

condicoes_iniciais = [1, 0]
x = np.linspace(0, 10, 100)

solucao = solve_ivp(
    equacao,
    [0, 10],
    condicoes_iniciais,
    t_eval = x
)

print("Valores de x: ", solucao.t)
print("Valores de y: ", solucao.y[0])
```

```python
# 12. Verifique quais colunas de um DataFrame contem valores duplicados.
import pandas as pd

dados = { "Produto": ["Mouse", "Teclado", "Mouse", "Monitor", "Teclado"],
      "Preco": [59, 98, 61, 800, 99],
      "Quantidade": [10, 5, 10, 2, 5]
  }
df = pd.DataFrame(dados)

# o duplicated() vai verificar as se foi a primeira vez que o item apareceu ou não,
# e o any() vai verificar se existe algum True (que apareceu mais de uma vez)
for coluna in df.columns:
  if (df[coluna].duplicated().any()):
    print(f"A coluna {coluna} possui valores duplicados")

print()
print(df)
```

```python
# 13. Crie uma tabela dinamica a partir do DataFrame, calculando a soma de Quantidade por Categoria.
import pandas as pd

dados = { "Produto": ["Mouse", "Teclado", "Monitor", "Gabinete", "CPU"],
      "Categoria": ["Periférico", "Periférico", "Monitor", "Hardware", "Hardware"],
      "Quantidade": [10, 5, 2, 7, 3]
    }
df = pd.DataFrame(dados)

# usa o pivot_table() para dizer que vamos calcular a coluna quantidade, agrupando pela categoria fazendo a soma
tabela = df.pivot_table(values = 'Quantidade', index = 'Categoria', aggfunc = 'sum')

print(tabela)
```

```python
# 14. Mescle dois DataFrames baseados em uma coluna comum chamada ID utilizando um join.
import pandas as pd

clientes = pd.DataFrame({
    "ID": [1, 2, 3],
    "Nome": ["Matheus", "João", "Maria"]
})
compras = pd.DataFrame({
    "ID": [1, 2, 3],
    "Produto": ["Mouse", "Teclado", "Monitor"]
})

# cria um novo DF apartir de clientes juntando com compras seguindo o index ID
clientes_compras = clientes.join(compras.set_index("ID"), on = "ID")

print(clientes_compras)
```

```python
# 15. Crie um DataFrame com dados faltantes e use a interpolacao linear para preenche-los.
import pandas as np
import numpy as np

df = pd.DataFrame({'Produto': ['Mouse', 'Monitor', 'Teclado', 'Gabinete', 'CPU'],
      'Preco': [59, np.nan, 99, np.nan, 1200],
      'Quantidade': [10, 2, 5, 7, 8]
      })

# o interpolate() utiliza dos valores anteriores e posteriores ao NaN para determinar qual valor usar
df['Preco'] = df['Preco'].interpolate()

print(df)
```

```python
# 17. Adicione uma coluna ao DataFrame que calcule o imposto (5%) sobre o valor da coluna Total.
import pandas as pd

df = pd.DataFrame({ "Produto": ["Mouse", "Teclado", "Pendrive", "Monitor", "Mousepad"],
      "Preco": [59, 98, 61, 800, 99],
      "Quantidade": [10, 5, 10, 2, 5],
      "Total": [590, 490, 610, 1600, 495]
  })

# Cria a coluna imposto inserindo o resultado da conta
df['Imposto'] = df['Total'] * 0.05

print(df)
```

```python
# 19. Crie um DataFrame a partir de um dicionario de listas e calcule a media de cada coluna numerica.
import pandas as pd

dados = { "Produto": ["Mouse", "Teclado", "Pendrive", "Monitor", "Mousepad"],
      "Preco": [59, 98, 61, 800, 99],
      "Quantidade": [10, 5, 10, 2, 5],
      "Total": [590, 490, 610, 1600, 495]
  }

# cria o dataframe e calcula a media só aceitando colunas que sejam totalmente numéricas
df = pd.DataFrame(dados)
media = df.mean(numeric_only = True)

print(media)
```

```python
# 20. Agrupe os dados por uma coluna categorica e calcule a mediana de uma coluna numerica.
import pandas as pd

df = pd.DataFrame({ "Produto": ["Mouse", "Teclado", "Monitor", "Gabinete", "CPU"],
      "Categoria": ["Periférico", "Periférico", "Monitor", "Hardware", "Hardware"],
      "Quantidade": [10, 5, 2, 7, 3]
    })

# agrupa pela Categoria, pega a Quantidade e tira a mediana
df_filtrado = df.groupby("Categoria")["Quantidade"].median()

print(df)
print()
print(df_filtrado)
```

```python
# 1. Crie um grafico de linha para a funcao y = x3 no intervalo de x = −10 ate x = 10.
# importa o modulo pyplot da matplotlib e o numpy para gerar os dados
import matplotlib.pyplot as plt
import numpy as np

# Essa função vai criar 100 numeros espaçados de forma igual entre -10 e 10
x = np.linspace(-10, 10, 100)
y = x ** 3

# Aqui onde vai ser criado o grafico, editado seus rotulos e exibido
plt.plot(x, y)
plt.title("Gráfico de y = x3")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

```python
# 2. Plote um grafico de dispersao para 100 pontos aleatorios com uma paleta de cores baseada nos valores de y.
import matplotlib.pyplot as plt
import numpy as np

# Gera 100 pontos aleatórios para cada
x = np.random.randn(100)
y = np.random.randn(100)

# o scatter representa o grafico de dispersao. utilizamos o "c = y" (color) para fazer a cor depender do eixo y, o cmap = à colormap
plt.scatter(x, y, c = y, cmap = "cividis")
plt.title("Gráfico de Dispersão")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

```python
# 3. Gere um histograma para 500 numeros inteiros gerados aleatoriamente entre 0 e 100.
import matplotlib.pyplot as plt
import numpy as np

dados = np.random.randint(0, 101, 500)

# cria o histograma, com um intervalo nos dados de 10 e a cor entre eles de preta
plt.hist(dados, bins = 10, edgecolor = "black")
plt.title("Histograma")
plt.xlabel("Valores")
plt.ylabel("Frequencia")
plt.show()
```

```python
# 4. Crie um grafico de barras para 5 categorias com valores diferentes, colorindo cada barra de uma cor diferente.
import matplotlib.pyplot as plt

categorias = ["A", "B", "C", "D", "E"]
valores = [10, 21, 5, 7, 1]

# color vai seguir respectivamente os elementos de valores e categorias, para poder aplicar a cor ao seu devido elemento
plt.bar(categorias, valores, color = ["blue", "red", "yellow", "orange", "brown"])
plt.title("Gráfico de Barras")
plt.xlabel("Categorias")
plt.ylabel("Valores")
plt.show()
```

```python
# 5. Faca um grafico de pizza para representar as porcentagens de quatro setores de mercado com rotulos personalizados.
import matplotlib.pyplot as plt

setores = ["Frios", "Açouque", "Padaria", "Congelados"]
porcentagem = [20, 30, 40, 10]

# porcentagem = tamanho da fatia, setores são os labels das fatias, autopct mostra a % dentro de cada fatia
plt.pie(porcentagem, labels = setores, autopct = "%1.1f%%")
plt.title("Setores do Mercado")

plt.show()
```

```python
# 6. Plote dois graficos de linha em subplots diferentes: y1 = ex e y2 = log(x).
import matplotlib.pyplot as plt
import numpy as np

# o x começa por 0.1 devido o log() ser maior do que 0
x = np.linspace(0.1, 5, 100)

# Calcula as funções
y1 = np.exp(x)
y2 = np.log(x)

# Primeiro subplot, cria um grafico com 2 linhas, 1 coluna e avisa que é o primeiro grafico
plt.subplot(2, 1, 1)
plt.plot(x, y1)
plt.title("y = ex")

# Segundo subplot, cria um grafico com 2 linhas, 1 coluna e avisa que é o segundo grafico
plt.subplot(2, 1, 2)
plt.plot(x, y2)
plt.title("y = log(x)")

# Ajusta o espaço entre um grafico e outro
plt.tight_layout()

plt.show()
```

```python
# 7. Crie um grafico de dispersao com tamanho dos pontos variando de acordo com os valores de y.
import matplotlib.pyplot as plt
import numpy as np

x = np.random.rand(100)
y = np.random.randn(100)

# o s de size vai utilizar a variavel y para variar o tamanho dos pontos, multipliquei o valor para ficar mais visual a diferença de tamanho
plt.scatter(x, y, s = y * 100)
plt.title("Gráfico de dispersão")
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")

plt.show()
```

```python
# 8. Plote um grafico de linha com intervalos de confianca (sombramento) para a funcao y = sin(x).
import matplotlib.pyplot as plt
import numpy as np

# Cria 100 pontos entre 0 e 2pi e calcula o seno
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

# define os limites do intervalo de confianca
limite_inferior = y - 0.2
limite_superior = y + 0.2

plt.plot(x, y, label = "sin(x)")

# Configura o sombreamento do intervalo
plt.fill_between(
    x,
    limite_inferior,
    limite_superior,
    alpha = 0.3,      # transpaencia do preenchimento
    label = "Intervalo de confiança"
)

plt.title("Função seno com intervalo de confiança")
plt.xlabel("X")
plt.ylabel("Y")

# Cria a legenda baseado nos labels
plt.legend()

plt.show()
```
